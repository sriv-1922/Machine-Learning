```python
# Simple Hypothesis Space Search
# Goal: Find the best rule for predicting
# whether a person will play outside based on weather

# Training data
data = [
    ("Sunny", "Yes"),
    ("Rainy", "No"),
    ("Cloudy", "Yes"),
    ("Sunny", "Yes"),
    ("Rainy", "No")
]

# Hypothesis 1: Always predict Yes
def always_yes(weather):
    return "Yes"

# Hypothesis 2: Always predict No
def always_no(weather):
    return "No"

# Hypothesis 3: Predict Yes only for Sunny
def sunny_only(weather):
    if weather == "Sunny":
        return "Yes"
    else:
        return "No"

# Hypothesis 4: Predict Yes when weather is not Rainy
def not_rainy(weather):
    if weather != "Rainy":
        return "Yes"
    else:
        return "No"

# Store all hypotheses
hypotheses = {
    "Always Yes": always_yes,
    "Always No": always_no,
    "Sunny Only": sunny_only,
    "Not Rainy": not_rainy
}

best_hypothesis = None
best_accuracy = 0

# Test each hypothesis
for name, rule in hypotheses.items():

    correct = 0

    print("\nTesting Hypothesis:", name)

    for weather, actual_answer in data:

        predicted_answer = rule(weather)

        print(
            "Weather:", weather,
            "| Actual:", actual_answer,
            "| Predicted:", predicted_answer
        )

        if predicted_answer == actual_answer:
            correct += 1

    accuracy = correct / len(data)

    print("Correct Predictions:", correct)
    print("Accuracy:", accuracy * 100, "%")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_hypothesis = name

print("\nBest Hypothesis:", best_hypothesis)
print("Best Accuracy:", best_accuracy * 100, "%")
```
