```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Sample dataset
X = np.array([
    [1.0], [1.2], [1.4], [1.6], [1.8],
    [5.0], [5.2], [5.4], [5.6], [5.8]
])

# Create Gaussian Mixture Model
model = GaussianMixture(n_components=2, random_state=42)

# Train using EM algorithm
model.fit(X)

# Predict clusters
labels = model.predict(X)

# Display results
print("Data Points:")
print(X.ravel())

print("\nPredicted Groups:")
print(labels)

print("\nGroup Means:")
print(model.means_)

print("\nGroup Probabilities:")
print(model.weights_)

# Plot the results
plt.figure(figsize=(8, 4))

plt.scatter(X, np.zeros_like(X), c=labels, s=100, cmap='viridis')

# Plot cluster means
plt.scatter(model.means_, [0, 0], marker='X', s=250, color='red', label='Cluster Means')

plt.title("Expectation Maximization using Gaussian Mixture Model")
plt.xlabel("Data Points")
plt.yticks([])
plt.legend()

plt.show()
```
