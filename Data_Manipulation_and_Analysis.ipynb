```python
import numpy as np

data = np.array([
    [101, 450],
    [102, 700],
    [103, 550],
    [104, 900]
])

# Save as NumPy file
np.save("sales.npy", data)

# Save as CSV file
np.savetxt(
    "sales.csv",
    data,
    delimiter=",",
    fmt="%d",
    header="Product_ID,Sales",
    comments=""
)

print("Data saved successfully")

# Load NumPy file
loaded_data = np.load("sales.npy")

print("Loaded Data:")
print(loaded_data)
```
