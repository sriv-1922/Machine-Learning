```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
prices = rng.integers(100, 5001, 20)
final_prices = prices * 1.18

print("Original prices:", prices)
print("First 5:", prices[:5])
print("Last 5:", prices[-5:])
print("Prices after 18% GST:", np.round(final_prices, 2))
print("Above Rs. 2500:", prices[prices > 2500])
print("Between Rs. 1000 and Rs. 3000:", prices[(prices >= 1000) & (prices <= 3000)])
print("Minimum:", prices.min(), "Maximum:", prices.max())
print("Mean:", prices.mean(), "Median:", np.median(prices))
print("Variance:", prices.var(), "Standard deviation:", prices.std())
print("Total value:", prices.sum())
print("Count above average:", np.sum(prices > prices.mean()))

pd.DataFrame({"Product_ID": np.arange(1, 21), "Original_Price": prices,
              "Final_Price_With_GST": np.round(final_prices, 2)}).to_csv("products.csv", index=False)
print("File saved successfully: products.csv")
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
temperatures = rng.integers(20, 46, 30)

print("Original temperatures:", temperatures)
print("First 7 days:", temperatures[:7])
print("Day 10:", temperatures[9], "C")
print("Day 20:", temperatures[19], "C")
print("Above 35 C:", temperatures[temperatures > 35])
print("Between 25 C and 30 C:", temperatures[(temperatures >= 25) & (temperatures <= 30)])
print("Coolest:", temperatures.min(), "C | Hottest:", temperatures.max(), "C")
print("Average:", temperatures.mean(), "Median:", np.median(temperatures))
print("Variance:", temperatures.var(), "Standard deviation:", temperatures.std())
print("Days above monthly average:", np.sum(temperatures > temperatures.mean()))

pd.DataFrame({"Day": np.arange(1, 31), "Temperature_C": temperatures}).to_csv("temperature.csv", index=False)
print("File saved successfully: temperature.csv")
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
units = rng.integers(100, 801, 12)
bills = units * 6

print("Original consumption:", units)
print("First 3 months:", units[:3])
print("Last month:", units[-1])
print("Bills at Rs. 6 per unit:", bills)
print("Months above 500 units:", units[units > 500])
print("Minimum: Month", units.argmin() + 1, "=", units.min(), "units")
print("Maximum: Month", units.argmax() + 1, "=", units.max(), "units")
print("Total annual consumption:", units.sum())
print("Average:", units.mean(), "Variance:", units.var(), "Standard deviation:", units.std())
print("Months below average:", np.sum(units < units.mean()))

pd.DataFrame({"Month": np.arange(1, 13), "Consumption_Units": units,
              "Bill_Rs": bills}).to_csv("electricity.csv", index=False)
print("File saved successfully: electricity.csv")
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
production = rng.integers(50, 501, (4, 7))

print("Complete 2D array:\n", production)
print("Machine 1 production:", production[0])
print("Day 3 production of all machines:", production[:, 2])
print("Machine 3, Day 5:", production[2, 4])
print("Production values above 300:", production[production > 300])
print("Total for each machine:", production.sum(axis=1))
print("Total for each day:", production.sum(axis=0))
print("Overall minimum:", production.min(), "Overall maximum:", production.max())
print("Overall mean:", production.mean(), "Overall standard deviation:", production.std())
print("Values above overall average:", np.sum(production > production.mean()))

pd.DataFrame(production, index=["Machine_1", "Machine_2", "Machine_3", "Machine_4"],
             columns=[f"Day_{day}" for day in range(1, 8)]).to_csv("factory_production.csv")
print("File saved successfully: factory_production.csv")
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
original_data = rng.integers(1, 1001, 50)
after_adding_100 = original_data + 100
final_data = after_adding_100 * 2

print("Original data:", original_data)
print("First 10:", original_data[:10])
print("Last 10:", original_data[-10:])
print("Every second element:", original_data[::2])
print("Indexes 5, 10, 15, 20:", original_data[[5, 10, 15, 20]])
print("After adding 100:", after_adding_100)
print("Final data after multiplying by 2:", final_data)
print("Greater than 500:", original_data[original_data > 500])
print("Between 200 and 600:", original_data[(original_data >= 200) & (original_data <= 600)])
print("Even numbers:", original_data[original_data % 2 == 0])
print("Odd numbers:", original_data[original_data % 2 != 0])
print("Sum:", original_data.sum())
print("Product:", np.prod(original_data, dtype=object))
print("Minimum:", original_data.min(), "Maximum:", original_data.max())
print("Mean:", original_data.mean(), "Median:", np.median(original_data))
print("Variance:", original_data.var(), "Standard deviation:", original_data.std())
print("25th, 50th, 75th percentiles:", np.percentile(original_data, [25, 50, 75]))
print("Count above mean:", np.sum(original_data > original_data.mean()))
print("Ascending sorted data:", np.sort(original_data))

np.save("original_data.npy", original_data)
pd.DataFrame({"Final_Processed_Data": final_data}).to_csv("final_data.csv", index=False)
print("File saved successfully: original_data.npy")
print("File saved successfully: final_data.csv")
```
