```python
# EM Algorithm using Gaussian Mixture Model

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs

# Generate sample data
X, y = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.0,
    random_state=42
)

# Create EM model
em = GaussianMixture(
    n_components=3,
    random_state=42
)

# Train the model
em.fit(X)

# Predict clusters
labels = em.predict(X)

# Display parameters
print("Means of clusters:")
print(em.means_)

print("\nCovariance matrices:")
print(em.covariances_)

print("\nMixing probabilities:")
print(em.weights_)

print("\nNumber of iterations:", em.n_iter_)
print("Log likelihood:", em.lower_bound_)

# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(
    em.means_[:, 0],
    em.means_[:, 1],
    marker='X',
    s=200
)

plt.title("Expectation-Maximization Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
